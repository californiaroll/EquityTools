```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

## Enter the tickers for the portfolio

```python
tickers = ['AAPL', 'NVDA', 'AMZN']
```

## Calculate start and end dates for 5 year range going back from today

```python
end_date = datetime.today()
```

```python
start_date = end_date - timedelta(days = 5 * 365)
```








## Download stock data

```python
data = yf.download( tickers, start = start_date, end = end_date)['Close']
```

## Enter the weights for the portfolio

```python
weights = np.array([0.3, 0.3, 0.4])
```

## Clean the data

```python
returns = data.pct_change().dropna()
```

## Now calculate the Sharpe Ratio

```python
portfolio_returns = (returns* weights).sum(axis=1)
```

```python
daily_mean = portfolio_returns.mean()
```

```python
daily_std = portfolio_returns.std()
```

```python
annualized_sharpe = (daily_mean / daily_std) * np.sqrt(252)
```

```python
print(f"Sharpe Ratio for this portfolio is {annualized_sharpe:.4f}")
```

## If you find a mutual fund or other investment with a Sharpe Ratio higher than 1.00, it’s worth taking a further look.

## If a good Sharpe Ratio is higher than 1.00, then a ratio of 1.50, 2.00, or higher is excellent.
