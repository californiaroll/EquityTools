# Merton Jump Diffusion Option Pricing in Python

```python
import numpy as np
from scipy.stats import norm
```

# Black-Scholes formulas

```python
def bs_call(S, K, T, r, sigma):
  d1 = (np.log(S/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  return S * norm.cdf(d1) - K * np.exp(-r*T) * norm.cdf(d2)
```

```python
def bs_put(S, K, T, r, sigma):
  d1 = (np.log(S/K) + (r + 0.5*sigma**2)*T) / (sigma*np.sqrt(T))
  d2 = d1 - sigma*np.sqrt(T)
  return K * np.exp(-r*T) * norm.cdf(-d2) - S * norm.cdf(-d1)
```

# Merton Jump Diffusion Call Pricing

```python
def factorial(n):
  if n == 0 or n == 1:
    return 1
  return n * factorial(n - 1)
```

```python
def merton_jump_call(S, K, T, r, sigma, m, v, lam, N=40):
  price = 0.0
  for k in range(N):
    r_k = r - lam*(m - 1) + (k * np.log(m)) / T
    sigma_k = np.sqrt(sigma**2 + (k * v**2) / T)
    poisson_prob = np.exp(-m*lam*T) * ((m*lam*T)**k) / factorial(k)
    price += poisson_prob * bs_call(S, K, T, r_k, sigma_k)
  return price
```

# Example usage

```python
S = 100 # Spot price
```

```python
K = 100 # Strike price
```

```python
T = 1 # Time to maturity (years)
```

```python
r = 0.02 # Risk-free rate
```

```python
sigma = 0.2 # Volatility
```

```python
m = np.exp(0 + 0.5*0.3**2) # Mean jump size adjustment
```

```python
v = 0.3 # Jump size volatility
```

```python
lam = 1 # Jump intensity
```

```python
call_price_jump_diffusion = round(merton_jump_call(S, K, T, r, sigma, m, v, lam), 4)
```

```python
print("Merton Jump Diffusion Call Price:", call_price_jump_diffusion)
```
