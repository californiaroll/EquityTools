```python
import yfinance as yf
```

```python
import pandas as pd
```

```python
import numpy as np
```

```python
ticker_symbol = "MSFT"
```

```python
start_date = '2025-01-01'
```

```python
end_date = '2026-02-02'
```

```python
stock_data = yf.download(ticker_symbol, start=start_date, end=end_date)
```

## Calculating Daily Returns

```python
stock_data['returns'] = stock_data['Close'].pct_change()
```

```python
print(stock_data)
```

```python
window_size = 5
```

```python
stock_data['volatility'] = stock_data['returns'].rolling(window=window_size).std() * np.sqrt(252) # Annualized volatility
```

```python
print(stock_data)
```

```python
import matplotlib.pyplot as plt
```

```python
# Line chart for daily returns
plt.figure(figsize=(10, 6))
plt.plot(stock_data.index, stock_data['volatility'])
plt.title('Stock Volatility')
plt.xlabel('Date')
plt.ylabel('Volatility')
plt.grid(True)
plt.show()
```

```python
# Line chart for daily returns
plt.figure(figsize=(10, 6))
plt.plot(stock_data.index, stock_data['Daily Return'])
plt.title('Stock Daily Returns')
plt.xlabel('Date')
plt.ylabel('Daily Return')
plt.grid(True)
plt.show()
```

```python
# Histogram of returns
plt.figure(figsize=(10, 6))
plt.hist(stock_data['returns'], bins=30, edgecolor='black')
plt.title('Distribution of Daily Returns')
plt.xlabel('Daily Return')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```
